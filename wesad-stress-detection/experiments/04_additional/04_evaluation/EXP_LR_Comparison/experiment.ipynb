# EXP-001 / EXP-002 / EXP-011 / EXP-016 — So sánh train–test

Đọc kết quả đã lưu, không train lại. Mã chung là đơn vị tổng hợp; `source_run_id` chỉ dùng truy vết. Điểm train là resubstitution, test là người ngoài train. Không dùng train để khẳng định tổng quát hóa.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Image
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src/wesad/models.py").is_file())
FOLDER = ROOT / "experiments/04_evaluation/EXP_LR_Comparison"
summary = pd.read_csv(FOLDER / "train_test_summary.csv")
metrics = pd.read_csv(FOLDER / "train_test_metrics.csv")
display(summary)
```

```python
for task, part in summary.groupby("task"):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    for ax, metric in zip(axes, ["macro_f1", "balanced_accuracy"]):
        part.set_index("experiment")[["train_" + metric + "_mean", "test_" + metric + "_mean"]].rename(columns={"train_" + metric + "_mean": "Train", "test_" + metric + "_mean": "Test"}).plot.bar(ax=ax, rot=0, ylim=(0, 1.05))
        ax.set_title(task + " — " + metric)
    fig.tight_layout()
    (FOLDER / "figures").mkdir(exist_ok=True)
    fig.savefig(FOLDER / "figures" / (task + "_train_test.png"), dpi=150)
    display(Image(filename=str(FOLDER / "figures" / (task + "_train_test.png"))))
    plt.close(fig)
```
