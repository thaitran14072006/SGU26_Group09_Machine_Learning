## 04 — protocol

15 outer LOSO folds cho từng task, train 14 người và test 1 người. multiclass: 1 baseline, 2 stress, 3 amusement.
binary: stress=1, còn lại=0. mỗi outer train có 3 inner GroupKFold để chọn tham số theo macro f1.
median imputation và StandardScaler fit bên trong từng train fold; không ROS.
outer test không dùng chọn tham số hoặc dừng epoch. mỗi lượt: 360 inner fits + 30 outer refits.
đánh giá bằng mean và sd mẫu (ddof=1) theo 15 người; không lấy điểm trên toàn bộ cửa sổ thay cho mean theo người.


```python
verification=read_json(run/'verification.json')
assert verification['status']=='passed' and verification['outer_folds']==30
print(json.dumps(verification,ensure_ascii=False,indent=2))
print('split ví dụ multiclass/S4:')
print(json.dumps(read_json(run/'splits.json')['multiclass/mlp/S4'],ensure_ascii=False,indent=2))
```

## bổ sung ngày 20/11/2026 — kiểm tra độ nhạy theo seed

người thực hiện: như. EXP chung: EXP-020. Câu hỏi: kết quả thay đổi thế nào khi đổi seed khởi tạo MLP? Giữ nguyên dữ liệu, lưới bốn ứng viên, outer LOSO, ba inner GroupKFold, max_iter 200 và không ROS. Seed 42 dùng kết quả chính đã có; seed 43 và 44 được chạy nested CV mới, mỗi seed đủ 30 outer fold và 360 inner fits. Ba seed được khai báo trước khi chạy bổ sung. Không chọn seed có điểm cao nhất làm kết quả chính.

```python
from pathlib import Path
import pandas as pd, json, sys
root = next(p for p in [Path.cwd(), Path.cwd()/'wesad-stress-detection', *Path.cwd().parents] if (p/'src/wesad').is_dir())
folder=root/'experiments/04_additional/EXP-020/seed_sensitivity/20261120'
print('ngày tổng hợp: 20/11/2026')
print('python:',sys.executable)
summary=pd.read_csv(folder/'summary_by_seed.csv')
print(summary.to_string(index=False))
```

Thanh sai số dưới đây là độ lệch chuẩn giữa 15 người tại từng seed. Đây không phải độ lệch chuẩn giữa các seed và không phải khoảng tin cậy.

```python
import matplotlib.pyplot as plt
summary=pd.read_csv(folder/'summary_by_seed.csv')
fig,axes=plt.subplots(1,2,figsize=(12,4.5),sharey=True)
for ax,task in zip(axes,['multiclass','binary']):
    g=summary[summary.task==task].sort_values('seed')
    ax.bar(g.seed.astype(str),g.macro_f1_mean,yerr=g.macro_f1_sd_subjects,capsize=5)
    ax.set(title=task,xlabel='seed',ylabel='macro f1',ylim=(0,1))
fig.tight_layout()
```

```python
scores=pd.read_csv(folder/'scores_by_subject_seed.csv')
print(pd.read_csv(folder/'variability_across_seeds.csv').to_string(index=False))
print(pd.read_csv(folder/'prediction_agreement.csv').to_string(index=False))
print('số outer fold mới:',len(scores[scores.seed.isin([43,44])]))
fig=plt.figure(figsize=(12,8))
plt.imshow(plt.imread(folder/'figures/subject_scores_by_seed.png'))
plt.axis('off')
fig.tight_layout()
```

### nhận xét

multiclass: Macro-F1 của seed 42/43/44 lần lượt là 0.6211 / 0.6145 / 0.6203. Khoảng biến động trung bình theo seed là 0.0065.

binary: Macro-F1 của seed 42/43/44 lần lượt là 0.8164 / 0.7927 / 0.8446. Khoảng biến động trung bình theo seed là 0.0520.

Điểm theo seed được tính từ cùng 15 người. Không gộp 45 lượt người-seed thành 45 người độc lập. Đây là kiểm tra độ nhạy của cả quá trình chọn tham số và huấn luyện MLP, không tách riêng ảnh hưởng khởi tạo với ảnh hưởng lựa chọn siêu tham số. Seed 42 vẫn là kết quả chính đã chốt. Các lượt bổ sung không tạo tập kiểm tra độc lập ngoài WESAD. Loss đã lưu là mục tiêu train, không phải test loss theo epoch.
